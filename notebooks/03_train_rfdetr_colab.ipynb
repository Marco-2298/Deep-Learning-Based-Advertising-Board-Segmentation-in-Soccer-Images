```python
from pathlib import Path
import json

DATASET_DIR = Path(
    "/content/drive/MyDrive/"
    "football-adboard-segmentation/datasets/rf_detr"
)

expected = {
    "train": (6196, 13349),
    "valid": (1328, 2858),
    "test": (1327, 2939),
}

for split_name, (expected_images, expected_annotations) in expected.items():

    annotations_path = (
        DATASET_DIR
        / split_name
        / "_annotations.coco.json"
    )

    assert annotations_path.exists(), (
        f"JSON mancante: {annotations_path}"
    )

    with annotations_path.open("r", encoding="utf-8") as f:
        coco = json.load(f)

    n_images = len(coco["images"])
    n_annotations = len(coco["annotations"])

    print(
        f"{split_name}: "
        f"{n_images} immagini, "
        f"{n_annotations} annotazioni"
    )

    assert n_images == expected_images
    assert n_annotations == expected_annotations

print("\n✓ Dataset COCO verificato correttamente")
```

```python
import pandas as pd

metrics_path = OUTPUT_DIR / "metrics.csv"

df = pd.read_csv(metrics_path)

print("File:", metrics_path)
print("Righe:", len(df))
print("\nColonne disponibili:")
print(df.columns.tolist())
```

```python
wanted_cols = [
    "epoch",
    "train/loss",
    "val/loss",
    "val/mAP_50_95",
    "val/mAP_50",
    "val/mAP_75",
    "val/mAR",
    "val/F1",
    "val/precision",
    "val/recall",
    "val/ema_mAP_50_95",
]

available_cols = [
    c for c in wanted_cols
    if c in df.columns
]

# Manteniamo le righe che contengono effettivamente metriche di validation
val_rows = df[
    df["val/mAP_50_95"].notna()
][available_cols].copy()

display(val_rows)
```

```python
metric = (
    "val/ema_mAP_50_95"
    if "val/ema_mAP_50_95" in df.columns
    else "val/mAP_50_95"
)

best_idx = df[metric].idxmax()

best = df.loc[best_idx]

print("MIGLIOR RISULTATO")
print("-----------------")
print("Epoca:", int(best["epoch"]))
print(f"{metric}: {best[metric]:.4f}")

for col in [
    "val/mAP_50_95",
    "val/mAP_50",
    "val/mAP_75",
    "val/mAR",
    "val/F1",
    "val/precision",
    "val/recall",
]:
    if col in df.columns and pd.notna(best[col]):
        print(f"{col}: {best[col]:.4f}")
```
